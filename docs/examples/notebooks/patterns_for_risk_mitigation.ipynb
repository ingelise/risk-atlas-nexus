# Risk Mitigation with CSIRO Responsible AI Patterns

This notebook demonstrates how the `ran-csiro-patterns` extension complements AI Atlas Nexus by:
1. **Identifying risks** in a use case (Atlas Nexus)

## Part 4: End-to-End Risk Identification → Mitigation

Now we'll use Atlas Nexus to identify risks in a use case, then surface applicable CSIRO patterns as mitigations.

### Setup Inference Engine

Atlas Nexus uses LLMs to identify risks. Configure an inference engine:

**Available engines**: Ollama, WML, RITS, vLLM

For this example, we'll use Ollama (local). See the [risk_to_ares_evaluation notebook](./risk_to_ares_evaluation.ipynb) for other engine configurations.

```python
from ai_atlas_nexus.blocks.inference import OllamaInferenceEngine
from ai_atlas_nexus.blocks.inference.params import (
    InferenceEngineCredentials,
    OllamaInferenceEngineParams,
)

# Configure Ollama (assumes running locally on localhost:11434)
inference_engine = OllamaInferenceEngine(
    model_name_or_path="granite3.3:8b",
    credentials=InferenceEngineCredentials(api_url="http://localhost:11434"),
    parameters=OllamaInferenceEngineParams(
        num_predict=1000, 
        num_ctx=8192, 
        temperature=0.2  # Low temperature for consistent risk identification
    ),
)

print("Inference engine configured: Ollama (Granite 3.3:8B)")
```

### Use Case: Content Recommendation System

Let's analyze risks for a real-world scenario: a content recommendation system that suggests articles to users based on reading history.

```python
# Create an Atlas Nexus instance
atlas = AIAtlasNexus()
print("Created AIAtlasNexus instance")
```

```python
# Define our use case
usecase = """
We are building a content recommendation system that:
- Analyzes user reading history and preferences
- Uses ML models to predict content interest
- Recommends articles, videos, and news from our platform
- Personalizes feeds for millions of users daily
- Operates in multiple countries with diverse regulatory requirements

Key system properties:
- User data includes demographics, reading history, and inferred interests
- Recommendations directly influence what content users see
- High-stakes: content distribution affects public discourse
- Real-time inference at scale
"""

print("Use Case: Content Recommendation System")
print(f"Description length: {len(usecase)} characters")
```

```python
# Identify risks using Atlas Nexus
print("Identifying risks from use case...\n")

risks = atlas.identify_risks_from_usecases(
    usecases=[usecase],
    inference_engine=inference_engine,
    taxonomy="ibm-risk-atlas",  # Use IBM Risk Atlas taxonomy
    zero_shot_only=False,  # Use chain-of-thought examples for better reasoning
    max_risk=15,  # Limit to top 15 risks
)[0]

print(f"Identified {len(risks)} risks:\n")
for i, risk in enumerate(risks, 1):
    print(f"{i}. {risk.name}")
    if hasattr(risk, 'description'):
        print(f"   Description: {risk.description}")
    print()
```

### Link Identified Risks to Mitigation Patterns

For each identified risk, find CSIRO patterns that address it.

```python
# Map risks to addressable principles
# This is a simplified mapping; in practice, you might use embeddings or LLMs for semantic matching

risk_principle_mapping = {
    'bias': 'fairness',
    'discrimination': 'fairness',
    'unfair': 'fairness',
    'transparency': 'transparency',
    'explainability': 'transparency',
    'black box': 'transparency',
    'interpretability': 'transparency',
    'accountability': 'accountability',
    'responsibility': 'accountability',
    'privacy': 'privacy',
    'data protection': 'privacy',
    'security': 'security',
    'robustness': 'robustness',
    'adversarial': 'robustness',
}

print("Principle mappings defined")
```

```python
# For each risk, find applicable patterns
risk_to_patterns = {}

for risk in risks:
    risk_name = risk.name.lower()
    applicable_patterns = set()  # Use set to avoid duplicates
    
    # Find relevant principles from the risk name
    for keyword, principle in risk_principle_mapping.items():
        if keyword in risk_name:
            patterns = get_patterns_for_principle(principle)
            applicable_patterns.update([p['id'] for p in patterns])
    
    if applicable_patterns:
        # Get full pattern objects
        pattern_objects = [p for p in all_patterns if p['id'] in applicable_patterns]
        risk_to_patterns[risk.name] = pattern_objects

print(f"Mapped {len(risk_to_patterns)} risks to mitigation patterns")
```

```python
# Display risk → pattern mappings
print("\n" + "="*80)
print("RISK IDENTIFICATION → MITIGATION PATTERNS")
print("="*80 + "\n")

for risk_name, patterns in sorted(risk_to_patterns.items()):
    print(f"\n🔴 Risk: {risk_name}")
    print(f"   Applicable Mitigation Patterns: {len(patterns)}")
    
    # Group patterns by type
    by_type = {}
    for p in patterns:
        ptype = p.get('patternType', 'unknown')
        if ptype not in by_type:
            by_type[ptype] = []
        by_type[ptype].append(p)
    
    for ptype, plist in sorted(by_type.items()):
        print(f"\n   {ptype.upper()} ({len(plist)}):")
        for p in plist[:3]:  # Show first 3 per type
            print(f"     • {p['name']}")
            if p.get('sourceUrl'):
                print(f"       → {p['sourceUrl']}")
        if len(plist) > 3:
            print(f"     ... and {len(plist) - 3} more")
```

## Part 5: Deep Dive — A Single Risk's Mitigation Strategy

Let's pick one identified risk and explore its complete mitigation landscape.

```python
# Pick the first risk for detailed analysis
if risks:
    focus_risk = risks[0]
    print(f"📌 Deep Dive: {focus_risk.name}")
    print(f"\nDescription: {focus_risk.description if hasattr(focus_risk, 'description') else 'N/A'}")
else:
    print("No risks identified")
```

```python
# Get ALL patterns and find those that could apply to this risk
# by searching across all principles, lifecycle stages, and stakeholders

print(f"\nSearching for applicable patterns...\n")

# Collect patterns across multiple dimensions
applicable_ids = set()

# By lifecycle stage
for stage in ['Design', 'Development', 'Deployment', 'Monitoring']:
    stage_patterns = get_patterns_for_lifecycle_stage(stage)
    if stage_patterns:
        print(f"  {stage}: {len(stage_patterns)} patterns")
        applicable_ids.update([p['id'] for p in stage_patterns])

# By governance focus (org-wide oversight)
governance = get_patterns_by_type('governance')
print(f"  Governance (org-wide): {len(governance)} patterns")
applicable_ids.update([p['id'] for p in governance])

# Get full objects
comprehensive_patterns = [p for p in all_patterns if p['id'] in applicable_ids]
print(f"\nTotal applicable patterns for {focus_risk.name}: {len(comprehensive_patterns)}")
```

```python
# Show how this risk should be mitigated across the AI lifecycle
print(f"\n📋 {focus_risk.name} → Lifecycle-Staged Mitigation\n")

lifecycle_stages = ['Design', 'Development', 'Deployment', 'Monitoring']

for stage in lifecycle_stages:
    stage_patterns = get_patterns_for_lifecycle_stage(stage)
    if stage_patterns:
        print(f"\n🔷 {stage.upper()} Phase")
        print(f"   Patterns: {len(stage_patterns)} available")
        print(f"   Recommended approach:")
        
        # Separate by type
        governance_p = [p for p in stage_patterns if p.get('patternType') == 'governance']
        process_p = [p for p in stage_patterns if p.get('patternType') == 'process']
        product_p = [p for p in stage_patterns if p.get('patternType') == 'product']
        
        if governance_p:
            print(f"   - Governance: {', '.join([p['name'] for p in governance_p[:2]])}")
        if process_p:
            print(f"   - Process: {', '.join([p['name'] for p in process_p[:2]])}")
        if product_p:
            print(f"   - Product: {', '.join([p['name'] for p in product_p[:2]])}")
```
